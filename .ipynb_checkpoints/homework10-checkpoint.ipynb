```python
df = df.set_index("Attribute").T
# t flips rows and columns (transpose)
df.index.name = "Planet"
df.reset_index(inplace=True)
# moves the index column into a regular one
df.columns.name = None
```

```python
# problem because can't perform math on strings, with instead js concatenate. 

# 2.1.3
skip_cols = ["Planet", "Ring System?", "Global Magnetic Field?"]

for col in df.columns:
    if col not in skip_cols:
        df[col] = pd.to_numeric(df[col], errors="coerce")

print(df.dtypes)
```

```python
# floats, integers, and objects meaning the "yes/no", etc. values
```

```python
def attach_units(df, old_col, new_col, unit):
    new_values = []
    for val in df[old_col]:
        q = float(val) * unit
        new_values.append(q)

    df[old_col] = new_values
    df.rename(columns={old_col: new_col}, inplace=True)

attach_units(df, "Diameter (km)", "Diameter (km)", u.km)
print(df["Diameter (km)"])
```

```python
# start working here tomorrow. 

peri_col = "Perihelion (10^6 km)"
aph_col  = "Aphelion (10^6 km)"
df[peri_col] = [ (v * 1e6) * u.km for v in df[peri_col] ]
df[aph_col]  = [ (v * 1e6) * u.km for v in df[aph_col] ]
sma = (df[peri_col] + df[aph_col]) / 2
sma_name = "Semi-Major Axis (km)"

aph_idx = df.columns.get_loc('Aphelion (10^6 km)')
df.insert(aph_idx + 1, sma_name, sma)
```

```python
# 12/2, orbital period to years. 
orbit_col = "Orbital Period (days)"
df[orbit_col] = [ (v / 365) * u.yr for v in df[orbit_col] ]
df.rename(columns={orbit_col: "Orbital Period (years)"}, inplace=True)
print(df.columns.tolist())

earth_period = df.loc[2, "Orbital Period (years)"]
print(f"Earth's Orbital Period: {earth_period:.4f}")
```

```python
print("An Astronomical Unit (AU) is the average distance between the Earth and the Sun.")
print("One AU =", const.au)
print("AU in kilometers:", const.au.to(u.km))
```

```python
distance_cols = [
    "Diameter (km)",
    "Distance from Sun (10^6 km)",
    "Perihelion (10^6 km)",
    "Aphelion (10^6 km)"
]

for col in distance_cols:
    df[col] = df[col].astype(str).str.replace(" km", "", regex=False)
    df[col] = pd.to_numeric(df[col], errors="coerce")
0
df["Diameter (km)"] = (u.Quantity(df["Diameter (km)"].values, u.km).to(u.AU).value)

million_cols = [
    "Distance from Sun (10^6 km)",
    "Perihelion (10^6 km)",
    "Aphelion (10^6 km)"
]

for col in million_cols:
    df[col] = u.Quantity(df[col].values * 1e6, u.km).to(u.AU).value



df = df.rename(columns={
    "Diameter (km)": "Diameter (AU)",
    "Distance from Sun (10^6 km)": "Distance from Sun (AU)",
    "Perihelion (10^6 km)": "Perihelion (AU)",
    "Aphelion (10^6 km)": "Aphelion (AU)"
})

planet = "Earth"
for col in df.columns:
    if "(AU)" in col:
        print(f"{planet} {col}: {df.loc[2, col]:.6f} AU")
```
