```python
import numpy as np
import matplotlib.pyplot as plt
# 2
def make_sine_wave(x, A, w):
    y = A * np.sin(w * x)
    return y

x = np.linspace(0, 2 * np.pi, 1000)

amplitude = [0.5, 1, 1.5, 2, 2.5]
frequency = [1, 2, 3, 4, 5]
linestyles = ["-", "--", "-.", ":", (0, (1, 2, 1, 2))]
# solid, dashed, dash dot, dotted, then custom ...
# of 1 line, 2 space, 1 line, 2 space, starting at position 0.

fig, axes = plt.subplots(1, 3, figsize=(25, 5))



for i, (A, w) in enumerate(zip(amplitude, frequency)):
    axes[0].plot(x, make_sine_wave(x, A, 1), linestyle=linestyles[i], label=f"A={A}, w=1")
    axes[1].plot(x, make_sine_wave(x, 1, w), linestyle=linestyles[i], label=f"A=1, w={w}")
    axes[2].plot(x, make_sine_wave(x, A, w), linestyle=linestyles[i], label=f"A={A}, w={w}")
    # enumerate adds the ability to assign an index value as well.


titles = [ 
    "Sine Waves with Different Amplitudes",
    "Sine Waves with Different Frequencies",
    "Sine Waves with Different Amplitudes and Frequencies"
]

for ax, title in zip(axes, titles):
    ax.set_title(title, fontsize=16)
    ax.set_xlabel("x", fontsize=14)
    ax.set_ylabel("y", fontsize=14)
    ax.legend(fontsize=12)
    ax.grid(True, alpha=0.3) # makes a kind of transparent grid

plt.tight_layout() # makes it so things don't overlap automatically
plt.show() # TUFFFFF
```
