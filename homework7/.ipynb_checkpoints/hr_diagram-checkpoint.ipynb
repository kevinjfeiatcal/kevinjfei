```python
import numpy as np
import matplotlib.pyplot as plt

data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
print(np.shape(data)) # 2642 rows, 2 columns. aka: 2642 stars, with brightness and temperature.

magnitude = data[:, 0]
temperature = data[:, 1]
print(np.min(magnitude), np.max(magnitude))
print(np.min(data), np.max(data))
brightest_index = np.argmin(magnitude)
dimmest_index = np.argmax(magnitude)
hottest_index = np.argmax(temperature)
coldest_index = np.argmin(temperature)


log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)

plt.figure(figsize=(8, 6))
plt.scatter(log_temperature, log_magnitude, color="gray", s=50, label="Stars")
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()

plt.scatter(log_temperature[brightest_index], log_magnitude[brightest_index],
            color="black", s=300, marker="*", label="Brightest Star")
plt.scatter(log_temperature[dimmest_index], log_magnitude[dimmest_index],
            color="purple", s=300, marker="*", label="Dimmest Star")
plt.scatter(log_temperature[hottest_index], log_magnitude[hottest_index],
            color="green", s=300, marker="*", label="Hottest Star")
plt.scatter(log_temperature[coldest_index], log_magnitude[coldest_index],
            color="orange", s=300, marker="*", label="Coldest Star")

plt.scatter(log_temperature, log_magnitude, c=temperature, cmap="inferno")
plt.colorbar(label="Temperature (K)")
plt.legend(fontsize=12)
plt.title("HR Diagram", fontsize=16)
plt.xlabel("Temperature (log Kelvin)", fontsize=14)
plt.ylabel("Magnitude (log nm)", fontsize=14)


```
