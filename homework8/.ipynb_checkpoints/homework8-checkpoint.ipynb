```python
median_lat = df["latitude"].median()
median_lon = df["longitude"].median()

df['abs_dist_lat_med'] = (df['latitude'] - median_lat).abs()
df['abs_dist_lon_med'] = (df['longitude'] - median_lon).abs()
df.shape

closest_lat_idx = df['abs_dist_lat_med'].idxmin()
farthest_lat_idx = df['abs_dist_lat_med'].idxmax()

closest_lon_idx = df['abs_dist_lon_med'].idxmin()
farthest_lon_idx = df['abs_dist_lon_med'].idxmax()

```

```python
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = sns.load_dataset("dowjones")
df = df.dropna()
```

```python
df.shape
```

```python
df.tail()
```

```python
df.head()
```

```python
print(df.columns)
```

```python
fig, axes = plt.subplots(2, 2, figsize=(12,10))

axes[0, 0].scatter(df["Date"], df["Price"], label="price", s=30, color="green")
axes[0, 0].set_title("Date v Price of DOW")
axes[0, 0].set_xlabel("Date (YYYY/MM/DD)")
axes[0, 0].set_ylabel("Price ($)")
axes[0, 0].legend()

axes[0, 1].scatter(df["Price"], df["Date"], label="date", s=30, color="red")
axes[0, 1].set_title("Price v Date of DOW")
axes[0, 1].set_xlabel("Price ($)")
axes[0, 1].set_ylabel("Date (YYYY/MM/DD)")
axes[0, 1].legend()

sns.histplot(df["Price"], bins = 30, ax=axes[1, 0], color="blue")
axes[1, 0].set_title("Histogram of DOW Prices by Date")
axes[1, 0].set_xlabel("Prices ($)")
axes[1, 0].set_ylabel("Date (YYYY/MM/DD)")
# issue- historgram doesn't show the intended way well, so had to switch
# the axes, but the funcitonality of the graph is lost. think line plot
# might work. 

sns.boxplot(y=df["Price"], ax=axes[1, 1], color="orange")
axes[1, 1].set_title("Boxplot of DOW Prices over Time")
axes[1, 1].set_ylabel("Price ($)")

plt.tight_layout()
plt.show() # why aint it showin...
```
